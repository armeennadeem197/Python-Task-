```python
# Question 1 — Student Attendance Manager
# Ek program banayein jo students ki attendance manage kare.
# Student name
# Student ID
# Attendance percentage
# Requirements:
# Students ko dictionary mein store karein.
# Tamam students ko list mein save karein.
# Student ID duplicate nahi honi chahiye.
# Student ID se student search karein.
# Attendance 75% ya above ho to Allowed display karein.
# 75% se kam ho to Short Attendance display karein.
# Total students count karein.
# Separate functions use karein.

students = []


# Add Student
def add_student():

    student_id = int(input("Enter Student ID: "))

    # Duplicate ID check
    for student in students:

        if student["id"] == student_id:
            print("Student ID already exists!")
            return

    name = input("Enter Student Name: ")
    attendance = float(input("Enter Attendance Percentage: "))

    student = {
        "id": student_id,
        "name": name,
        "attendance": attendance
    }

    students.append(student)

    print("Student added successfully!")


# Search Student
def search_student():

    student_id = int(input("Enter Student ID to search: "))

    for student in students:

        if student["id"] == student_id:

            print("\nStudent Found!")
            print("ID:", student["id"])
            print("Name:", student["name"])
            print("Attendance:", student["attendance"], "%")

            if student["attendance"] >= 75:
                print("Status: Allowed")
            else:
                print("Status: Short Attendance")

            return

    print("Student not found!")


# Display Students
def display_students():

    if len(students) == 0:
        print("No students available.")
        return

    for student in students:

        print("\nID:", student["id"])
        print("Name:", student["name"])
        print("Attendance:", student["attendance"], "%")

        if student["attendance"] >= 75:
            print("Status: Allowed")
        else:
            print("Status: Short Attendance")


# Main Menu
while True:

    print("\n===== Student Attendance Manager =====")
    print("1. Add Student")
    print("2. Search Student")
    print("3. Display Students")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_student()

    elif choice == "2":
        search_student()

    elif choice == "3":
        display_students()

    elif choice == "4":
        print("Program Ended!")
        break

    else:
        print("Invalid choice!")
```

```python
# Question 2 — Simple Bank Account Manager

# Ek program banayein jo bank accounts manage kare.

# Account Information:

# Account number
# Customer name
# Balance

# Requirements:

# Account ko dictionary mein store karein.
# Tamam accounts ko list mein save karein.
# Account number unique hona chahiye.
# Account number se search karein.
# Deposit function banayein.
# Withdraw function banayein.
# Agar balance se zyada amount withdraw karein to Insufficient Balance show karein.
# Account balance display karein.
# Separate functions use karein.

# Menu:

# 1. Add Account
# 2. Search Account
# 3. Deposit
# 4. Withdraw
# 5. Display Accounts
# 6. Exit

accounts = []


# Add Account
def add_account():

    account_number = int(input("Enter Account Number: "))

    # Duplicate Account Number check
    for account in accounts:

        if account["number"] == account_number:
            print("Account number already exists!")
            return

    name = input("Enter Customer Name: ")
    balance = float(input("Enter Initial Balance: "))

    account = {
        "number": account_number,
        "name": name,
        "balance": balance
    }

    accounts.append(account)

    print("Account added successfully!")


# Search Account
def search_account():

    account_number = int(input("Enter Account Number: "))

    for account in accounts:

        if account["number"] == account_number:

            print("\nAccount Found!")
            print("Account Number:", account["number"])
            print("Customer Name:", account["name"])
            print("Balance: Rs.", account["balance"])

            return

    print("Account not found!")


# Deposit
def deposit():

    account_number = int(input("Enter Account Number: "))

    for account in accounts:

        if account["number"] == account_number:

            amount = float(input("Enter Deposit Amount: "))

            account["balance"] = account["balance"] + amount

            print("Amount deposited successfully!")
            print("New Balance: Rs.", account["balance"])

            return

    print("Account not found!")


# Withdraw
def withdraw():

    account_number = int(input("Enter Account Number: "))

    for account in accounts:

        if account["number"] == account_number:

            amount = float(input("Enter Withdraw Amount: "))

            if amount <= account["balance"]:

                account["balance"] = account["balance"] - amount

                print("Amount withdrawn successfully!")
                print("Remaining Balance: Rs.", account["balance"])

            else:
                print("Insufficient Balance!")

            return

    print("Account not found!")


# Display Accounts
def display_accounts():

    if len(accounts) == 0:
        print("No accounts available.")
        return

    for account in accounts:

        print("\nAccount Number:", account["number"])
        print("Customer Name:", account["name"])
        print("Balance: Rs.", account["balance"])


# Main Menu
while True:

    print("\n===== Bank Account Manager =====")
    print("1. Add Account")
    print("2. Search Account")
    print("3. Deposit")
    print("4. Withdraw")
    print("5. Display Accounts")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_account()

    elif choice == "2":
        search_account()

    elif choice == "3":
        deposit()

    elif choice == "4":
        withdraw()

    elif choice == "5":
        display_accounts()

    elif choice == "6":
        print("Thank you!")
        break

    else:
        print("Invalid choice!")
```
